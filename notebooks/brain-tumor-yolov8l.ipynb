```python
#Création de fichier yaml qui décrit la structure du dataset pour entraîner le modèle YOLO.
import yaml

dataset_yaml={
    'path':r'/kaggle/input/brain-tumor-s640/Brain_Tumor_4CL_copy_resizing',
    'train':'Train',
    'val':'Valid',
    'nc': 4,
    'names':["Glioma","Meningioma","No Tumor","Pituitary"]
}

with open(r'/kaggle/working/dataset_yaml.yaml', 'w') as file:
    yaml.dump(dataset_yaml, file)

print("fichier YAML créé.")
```

```python
from ultralytics import YOLO

#Chargement de modèle YOLOv8 et l'entraîner sur dataset
model=YOLO("yolov8l.pt")
results_train=model.train(data=r"/kaggle/working/dataset_yaml.yaml",epochs=100,imgsz=640)
```

```python
#Enregistrer le modèle YOLOv8
model.save('/kaggle/working/B_yolov8l.pt')
```
